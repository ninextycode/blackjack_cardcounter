```python
from android_capture import ScreenCapture
import time
import image_utils
import get_cards_playing
import ocr_cards
import matplotlib.pyplot as plt
from get_cards_playing import (
    get_player_cards,
    get_initial_dealer_cards,
    get_final_dealer_cards
)
```

```python
capturer = ScreenCapture()
```

```python
im = capturer.get_screen()
```

```python
f, ax = plt.subplots(1, 1, figsize=(8, 4))
ax.imshow(im)
ax.set_axis_off()
```

```python
print("player_cards, initial_dealer_cards, final_dealer_cards")


while True:
    im = capturer.get_screen()

    player_cards = get_player_cards(im)
    initial_dealer_cards = get_initial_dealer_cards(im)
    final_dealer_cards = get_final_dealer_cards(im)

    if len(player_cards) > 0 or len(initial_dealer_cards) > 0 or len(final_dealer_cards) > 0:
        print(player_cards, initial_dealer_cards, final_dealer_cards)

    time.sleep(0.2)
```

```python
f, ax = plt.subplots(1, 1, figsize=(8, 4))
ax.imshow(im)
ax.set_axis_off()
```
