```python
wrapped_shoe_start = image_utils.wrap_perspective(img_np_start, poly, dest_w=100, dest_h=201)
wrapped_shoe_end = image_utils.wrap_perspective(img_np_end, poly, dest_w=100, dest_h=201)

f, [ax0, ax1] = plt.subplots(1, 2, figsize=(8, 6))
f.tight_layout()
ax0.imshow(wrapped_shoe_start, aspect='equal')
ax1.imshow(wrapped_shoe_end, aspect='equal')
```

```python
gray_start = cv2.cvtColor(wrapped_shoe_start, cv2.COLOR_RGB2GRAY)
gray_end = cv2.cvtColor(wrapped_shoe_end, cv2.COLOR_RGB2GRAY)

hsv_start = cv2.cvtColor(wrapped_shoe_start, cv2.COLOR_RGB2HSV)
hsv_end = cv2.cvtColor(wrapped_shoe_end, cv2.COLOR_RGB2HSV)


def get_white_mask(img_rgb):
    hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
    return get_white_mask_hsv(hsv)

def get_white_mask_hsv(hsv):
    lower_white = np.array([0, 0, 200])
    upper_white = np.array([180, 30, 255])
    mask = cv2.inRange(hsv, lower_white, upper_white)
    return mask > 0


f, [ax0, ax1] = plt.subplots(1, 2, figsize=(8, 6))
f.tight_layout()

ax0.imshow(get_white_mask_hsv(hsv_start), aspect='equal')
ax1.imshow(get_white_mask_hsv(hsv_end), aspect='equal')
```

```python
white_mask_0 = get_white_mask(wrapped_shoe_start[:, :20])
white_mask_1 = get_white_mask(wrapped_shoe_end[:, :20])
```

```python
white_mean_0 = white_mask_0.mean(axis=1)
white_mean_1 = white_mask_1.mean(axis=1)
```

```python
f, ax = plt.subplots(1, 1, figsize=(8, 6))
ax.plot(white_mean_0, label='start')
ax.plot(white_mean_1, label='end')
```

```python
def get_first_white_line_idx(white_sum, threshold=0.9):
    for idx, val in enumerate(white_sum):
        if val > threshold:
            return idx
    return None
```

```python
idx_0 = get_first_white_line_idx(white_mean_0)
idx_1 = get_first_white_line_idx(white_mean_1)
```
